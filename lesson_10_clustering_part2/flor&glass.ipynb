```python
def solve_glass_problem(glasses, floors):
    # створюємо таблицю, де будемо зберігати макс. висоту, яку можемо перевірити
    # рядки—спроби, стовп—стакани
    dp = [[0] * (glasses + 1) for _ in range(floors + 1)]
    
    attempts = 0
    #поки за поточну кількість спроб не зможемо покрити всі поверхи
    while dp[attempts][glasses] < floors:
        attempts += 1
        for g in range(1, glasses + 1):
            # якщо розбився дивимось скільки поверхів нижче можна б перевірити (спроб і стаканів стає на -1)
            # якщо не розбився дивимось скільки можна перевірити вище (стаканів так само спроб на -1)
            # +1 — це той поверх, з якого кинули зараз
            dp[attempts][g] = dp[attempts - 1][g - 1] + dp[attempts - 1][g] + 1
            
    # вертаємо кількість спроб, на якій ми нарешті "добилися" до потрібного поверху
    return attempts

# наші умови задачі
tasks = [
    (2, 10), 
    (2, 100), 
    (3, 100), 
    (3, 1000), 
    (5, 1000)
]

for g, f in tasks:
    ans = solve_glass_problem(g, f)
    print(f"Поверхів: {f:4} | Стаканів: {g} | Спроб: {ans}")
```
