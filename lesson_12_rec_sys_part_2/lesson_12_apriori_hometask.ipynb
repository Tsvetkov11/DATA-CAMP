```python
import pandas as pd
df = pd.read_csv('Groceries_dataset.csv')
print(df.head())
print(f"Загалом: {len(df)}")
```

```python
transactions = df.groupby(['Member_number', 'Date'])['itemDescription'].apply(list).tolist()
print(transactions[:3])
```

```python
from apyori import apriori
rules = apriori(transactions, 
                min_support=0.001,   
                min_confidence=0.05, 
                min_lift=1.2,
                min_length=2)        

results = list(rules)
results.sort(key=lambda x: x.ordered_statistics[0].lift, reverse=True)

for i in range(10):
    print(f"Rule #{i+1}: {list(results[i].items)}")
    print(f"Lift: {results[i].ordered_statistics[0].lift:.2f}")
```
