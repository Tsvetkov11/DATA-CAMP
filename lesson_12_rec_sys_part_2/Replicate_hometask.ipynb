```python
from scipy.io import loadmat
import numpy as np
mat = loadmat('ex8_movies.mat')
Y = mat['Y']
R = mat['R']

movie_means = np.sum(Y, axis=1) / (np.sum(R, axis=1) + 1e-9)
top_indices = np.argsort(movie_means)[::-1]

print("top 10 Simple Recommendations:")
for i in range(10):
    idx = top_indices[i]
    print(f"#{i+1}: Movie ID {idx+1} | Average Rating: {movie_means[idx]:.2f}")
```

```python
import numpy as np
my_ratings = np.zeros((1682, 1))
my_ratings[0] = 4
my_ratings[11] = 5
my_ratings[182] = 4
```

```python
params = loadmat('ex8_movieParams.mat')
X = params['X'] 
user_profile = np.dot(X.T, my_ratings) / (np.sum(my_ratings) + 1e-9)

content_predictions = np.dot(X, user_profile)
content_top_indices = np.argsort(content_predictions.flatten())[::-1]

print("\nT]op 5 Content-based Recommendations:")
for i in range(5):
    idx = content_top_indices[i]
    print(f"#{i+1}: Movie ID {idx+1}")
```
