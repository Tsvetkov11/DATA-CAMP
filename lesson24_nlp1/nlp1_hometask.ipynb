```python
from nltk.corpus import gutenberg
moby_raw = gutenberg.raw('melville-moby_dick.txt')
```

```python
#відношення унікальних токенів до їх загальної кількост
def answer_one():
    import nltk
    tokens = nltk.word_tokenize(moby_raw)
    return len(set(tokens)) / len(tokens)
```

```python
answer_one()
```

```python
#відсоток входження конкретних слів (з великої та малої літери)
from nltk.tokenize import word_tokenize
def answer_two():
    tokens = word_tokenize(moby_raw)
    whale_count = tokens.count('whale') + tokens.count('Whale')
    return (whale_count / len(tokens)) * 100
```

```python
answer_two()
```

```python
#FreqDist для підрахунку частоти токенів.
def answer_three():
    import nltk
    from nltk.probability import FreqDist
    tokens = nltk.word_tokenize(moby_raw)
    dist = FreqDist(tokens)
    return dist.most_common(20)
```

```python
answer_three()
```

```python
#Фільтруємо словник частот за двома умовами і сортуємо за алфавітом
def answer_four():
    import nltk
    from nltk.probability import FreqDist
    tokens = nltk.word_tokenize(moby_raw)
    dist = FreqDist(tokens)
    # ВІдбираємо що відповідають умовам
    ans = [word for word in dist.keys() if len(word) > 5 and dist[word] > 150]
    return sorted(ans)
```

```python
answer_four()
```

```python
#шукаємо найдовший токен за допомогою max
def answer_five():
    import nltk
    tokens = nltk.word_tokenize(moby_raw)
    #найдовше слово використовуючи len як ключ
    longest_word = max(tokens, key=len)
    return (longest_word, len(longest_word))
```

```python
answer_five()
```

```python
#відфільтрувати пунктуацію, .isalpha()

def answer_six():
    import nltk
    from nltk.probability import FreqDist
    tokens = nltk.word_tokenize(moby_raw)
    dist = FreqDist(tokens)
    # частота > 2000 і це має бути слово
    ans = [(freq, word) for word, freq in dist.items() if freq > 2000 and word.isalpha()]
    #спадання частоти
    return sorted(ans, key=lambda x: x[0], reverse=True)
```

```python
 answer_six()
```

```python
#розбиваємо текст на речення і на токени+ділимо одне на інше

from nltk.tokenize import sent_tokenize, word_tokenize
import numpy as np

def answer_seven():
    sentences = sent_tokenize(moby_raw)
    tokens = word_tokenize(moby_raw)
    return len(tokens) / len(sentences)
```

```python
answer_seven()
```

```python
#pos_tag для визначення частин мови, а потім рахуємо їхню частоту
def answer_eight():
    import nltk
    from nltk.probability import FreqDist
    tokens = nltk.word_tokenize(moby_raw)
    pos_tags = nltk.pos_tag(tokens)

    pos_counts = FreqDist([tag for word, tag in pos_tags])
    return pos_counts.most_common(5)
```

```python
answer_eight()
```

```python
#правильне слово за мінімальною відстанню(edit_distance)
from nltk.corpus import words
import nltk

def answer_nine(default_words=['cormulent', 'incendenece', 'validrate']):
    correct_spellings = words.words()
    recommendations = []
    
    for word in default_words:
        candidates = [w for w in correct_spellings if w[0] == word[0]]
        best_match = min(candidates, key=lambda x: nltk.edit_distance(word, x, transpositions=True))
        recommendations.append(best_match)
        
    return recommendations
```

```python
answer_nine()
```
