```python
import pandas as pd
from sklearn.model_selection import train_test_split

fn_neg = 'rt-polarity.neg'
with open(fn_neg, "r", encoding='utf-8', errors='ignore') as f:
    texts_neg = f.read().splitlines()

fn_pos = 'rt-polarity.pos'
with open(fn_pos, "r", encoding='utf-8', errors='ignore') as f:
    texts_pos = f.read().splitlines()

df_neg = pd.DataFrame({'text': texts_neg, 'label': 0})
df_pos = pd.DataFrame({'text': texts_pos, 'label': 1})

#одиндатасет
df = pd.concat([df_neg, df_pos], ignore_index=True)

#TrainTest
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], 
    df['label'], 
    test_size=0.2, 
    random_state=42, 
    stratify=df['label']
)

print(f"Кількість рядків у навчальній вибірці (Train): {len(X_train)}")
print(f"Кількість рядків у тестовій вибірці (Test): {len(X_test)}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

#векторизація TF-IDF
tfidf_vect = TfidfVectorizer(max_features=5000) 
X_train_tfidf = tfidf_vect.fit_transform(X_train)
X_test_tfidf = tfidf_vect.transform(X_test)

#Logistic Regression
model_1 = LogisticRegression(max_iter=1000, random_state=42)
model_1.fit(X_train_tfidf, y_train)


y_pred_1 = model_1.predict(X_test_tfidf)

#метрики
metrics_1 = {
    'Accuracy': accuracy_score(y_test, y_pred_1),
    'Precision': precision_score(y_test, y_pred_1),
    'Recall': recall_score(y_test, y_pred_1),
    'F1-score': f1_score(y_test, y_pred_1)
}

print("TF-IDF + Logistic Regression")
for metric_name, value in metrics_1.items():
    print(f"{metric_name}: {value:.4f}")
```

```python
tfidf_vect_ngrams = TfidfVectorizer(ngram_range=(1, 2), max_features=10000) 
X_train_ngrams = tfidf_vect_ngrams.fit_transform(X_train)
X_test_ngrams = tfidf_vect_ngrams.transform(X_test)

model_2 = LogisticRegression(max_iter=1000, random_state=42)
model_2.fit(X_train_ngrams, y_train)

y_pred_2 = model_2.predict(X_test_ngrams)

#метрики
metrics_2 = {
    'Accuracy': accuracy_score(y_test, y_pred_2),
    'Precision': precision_score(y_test, y_pred_2),
    'Recall': recall_score(y_test, y_pred_2),
    'F1-score': f1_score(y_test, y_pred_2)
}

print("TF-IDF Bigrams + Logistic Regression")
for metric_name, value in metrics_2.items():
    print(f"{metric_name}: {value:.4f}")
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

#наївни1 Баєс 
model_3 = MultinomialNB(alpha=1.0) 
model_3.fit(X_train_tfidf, y_train)

y_pred_3 = model_3.predict(X_test_tfidf)

metrics_3 = {
    'Accuracy': accuracy_score(y_test, y_pred_3),
    'Precision': precision_score(y_test, y_pred_3),
    'Recall': recall_score(y_test, y_pred_3),
    'F1-score': f1_score(y_test, y_pred_3)
}

print("Naive Bayes")
for metric_name, value in metrics_3.items():
    print(f"{metric_name}: {value:.4f}")
```

```python
import pandas as pd

data_comparison = {
    'Method 1: TF-IDF + Logistic Regression': metrics_1,
    'Method 2: TF-IDF Bigrams + Logistic Regression': metrics_2,
    'Method 3: TF-IDF + Naive Bayes': metrics_3
}

df_results = pd.DataFrame(data_comparison).T

df_results
```
